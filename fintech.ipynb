```python
import pandas as pd
import numpy as np

df=pd.read_excel("Payment_transaction.xlsx") #read the excel
df.head(5) #first 5rows display
```

```python
df.info() #data details 
```

```python
df.describe(include="all") #describe the all datatype
```

```python
df.columns=(df.columns.str.strip().str.lower().str.replace(" ","_")) #column name modify 
df.head(5)
```

```python
#check the null and negelect the null values 
df.isnull().sum()
df=df.dropna(subset =["customer_id","transaction_date","transaction_time","payment_method","bank_name","city"])
df.isnull().sum()
```

```python
#check ths duplicates
for col in df.columns:
    print(col, df[col].duplicated().sum())
```

```python
#drop the duplicates values
df.drop_duplicates(subset=["transaction_id","customer_id"],inplace=True) #while assigning don't use the inplace it will show error 
for col in df.columns:
    print(col, df[col].duplicated().sum())
```

```python
#remove the space in data of all the column
for col in df.select_dtypes(include=["object","number"]).columns:
   df[col]=df[col].str.strip().str.title()

df.head(5)
```

```python
#cleaning transaction amount
df["transaction_amount"]=(df["transaction_amount"].str.replace("₹","").
                          str.replace(",","").
                          str.replace("Inr","").
                          str.strip())

```

```python
df["transaction_amount"]=pd.to_numeric(df["transaction_amount"],errors="coerce")
df.info()
df=df[df["transaction_amount"]>0]
df.head(5)
```

```python
#cleaning date format
df["transaction_date"]=pd.to_datetime(df["transaction_date"],errors="coerce",dayfirst=True,format="mixed")
df = df.dropna(subset=["transaction_date"])
df.head(5)
```

```python
#cleaning time format
df["transaction_time"]=(pd.to_datetime(df["transaction_time"].str.replace(".",":"),errors="coerce",format="mixed")).dt.strftime("%H:%M")
df["transaction_time"].isnull().sum()
df.head(5)
```

```python
# I need to merge the twocolumns
```

```python
df["transaction_datetime"]=df["transaction_date"].astype(str).str.strip()+" " +df["transaction_time"].astype(str).str.strip()
df.drop(columns=["transaction_date","transaction_time"], inplace=True)
df.head(5)

```

```python
df.insert(
    2,
    "transaction_datetime",
    df.pop("transaction_datetime")
)
```

```python
df.head(5)
```

```python
#replace the value in the payment methods

changing={
    "Cc":"Credit Card",
    "CreditCard":"Credit Card",
    "Debitcard":"Debit Card",
    "Upi":"UPI",
    "Netbanking":"Net Banking",
    "Wallet":"Wallet"
}

df["payment_method"]=df["payment_method"].replace(changing)  # we can use map and replace if you give all he value replace go with map some value changing go with replace 
df.head(5)
```

```python
ch={"Sbi":"SBI",
   "Kotak Mahindra Bank":"Kotak Bank"}
df["bank_name"]=df["bank_name"].replace(ch)
df.head(5)
```

```python
chi={"Bengaluru":"Bangalore"}
df["city"]=df["city"].replace(chi)
df.head(5)
```

```python
wn={"Ios":"IOS"}
df["device_type"]=df["device_type"].replace(wn)

df.head(5)
```

```python
df["failure_reason"]=df["failure_reason"].fillna("Unknown")
```

```python
df.head(5)
```

```python
df.to_excel("final.xlsx",index=False) # excel save 
```
